
## The significance of Structured Streaming
**Spark Structured Streaming 实时数据流控制宇宙！**

推荐你在敲下第一行流式算子之前，先明白Structured Streaming实时监控技术在工业界被设计出来的**最高物理救赎意义**：

在旧时代，大数据是以天/小时为单位的“死水”（批处理 Batch）。但随着高频量化、实时风控、大厂秒级大盘的到来，数据变成了“一秒钟都在奔跑的活水（数据流 Stream）


旧时代的实时流框架（如早期的 Storm/Spark Streaming DStream）极其难用，因为它们把流数据看作离散的积木块，写起代码来和普通的批处理完全是两套外星语言。

**现代 Structured Streaming 彻底颠覆了这种割裂，提出了最伟大的物理概念：【把奔跑的数据流，看作一张在时空里无限追加、没有尽头的虚拟大表（Unbounded Table）写流处理，和写普通的 SQL / DataFrame 没有任何区别。

---

## UNbounded Table and Micro-Batch

#### 1. 无界Unbounded Table
新来的流式数据就像不停追加的行，永远在向这张表的底部插数据

#### 2. 微批次Micro-Batch触发器
Spark并不是来一条数据就惊动一次CPU。它极其聪明，会在底层设置一个“时间闹钟”（比如每 2 秒）。在这 2 秒内网线里进来的所有数据，会被打包成一个小小的 Micro-Batch。CPU 轰鸣一次，把这个小批次算完，然后继续等下一个 2 秒。这就是**高吞吐、低延迟的物理平衡**。

---
## Summarize
只要你在写Structured Streaming代码，你每时每刻都在用Unbounded table
在大数据体系里，Unbounded Table是一种抽象概念（Programming Model）

什么时候用 Micro-Batch？当你在落盘、在触发计算时，通过设置 Trigger 来决定用不用它、用什么频率用它。
